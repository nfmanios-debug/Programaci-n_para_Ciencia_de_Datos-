En este taller revisaremos las opciones que nos ofrece **pandas** para crear
agrupación y funciones de agregación y resumen sobre dataframes.

```python
# Vamos a seguir trabajando con el archivo de personas, así que en esta celda
# Importar Pandas y guarda en la variable df el archivo Excel
# ubicado en: https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx
import pandas as pd
url = 'https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx'
df = pd.read_excel(url)
df.head()
```

```python
# Para recordar un poquito
# Determinar el promedio del peso de las personas que no fuman y que tengan 2 o 3 hijos

df1 = df.loc[
    (df['Fuma'] == 'NO') &
    (df['No de hijos'].between(2, 3)),
    'Peso'].mean()
display(df1)
```

```python
# Ejemplo 1: Agrupar por una columna
# Promedio de edad por cada estrato socioeconómico
df2 = df.groupby('Estrato Socio')['Edad'].mean()
display(df2)
```

```python
# Ejemplo 2: Agrupar por una columna
# Por cada nivel educativo, determinar cuál es el ingreso más alto

df3 = df.groupby('Nivel Educativo')['Ingresos'].max().sort_values()
display(df3)

```

```python
# Ejemplo 3: Agrupar por varias  columnas
# Calcular el promedio de edad por género y número de hijos
df4 = df.groupby(['Genero','No de hijos'])['Ingresos'].max().sort_values()
display(df4)

```

```python
# Ejemplo 4: Agrupar por varias columnas
# Calcular la suma de ingresos por nivel educativo y estrato socioeconómico

df5 = df.groupby(['Nivel Educativo','Estrato Socio'])['Ingresos'].sum()
display(df5)
```

```python
# Ejemplo 5: Múltiples estadísticas
# Halle el promedio y la suma de edad agrupadas por la columna Fuma
df6 = df.groupby('Fuma').agg(
    Promedio_Edad=('Edad', 'mean'),
    Suma_Edad=('Edad', 'sum')
)
display(df6)
```

```python
# Ejemplo 6: Múltiples agregaciones
# Halle la suma, mayor y promedio de ingresos de las personas agrupadas género y si tiene casa
df6 = df.groupby('Fuma').agg(
    Promedio_Edad=('Edad', 'mean'),
    Suma_Edad=('Edad', 'sum')
)
display(df6)
```

```python
# Ejemplo 7: Múltiples agregaciones
# Halle la suma de los ingresos, el promedio del peso y el peso máximo de las personas agrupadas por nivel educativo

df7 = df.groupby('Nivel Educativo').agg(
    suma_ingresos=('Ingresos', 'sum'),
    Promedio_Peso=('Peso', 'mean'),
    Peso_Maximo=('Edad', 'max')
)
display(df7)

```

```python
# Ejemplo 8: Contar con groupby
# Cuántas personas hay en cada nivel educativo
df8 = df.groupby('Nivel Educativo').agg(
    total_personas=('Nivel Educativo', 'size')
)
display(df8)
```

```python
# Ejemplo 9: Contar con groupby
# Si agrupamos por las columnas tiene casa y tiene automóvil, determinar cuántas personas hay

df9 = df.groupby(['Tiene Casa','Tiene Automovil']).size()
display(df9)
```

```python
# Ejemplo 10: Groupby y filtros
# Ahora solo queremos trabajar con personas que tengan entre 30 y 40 años de edad,
# Y de ese grupo, determine cuántas personas hay en cada nivel educativo y
# cuál es el promedio de los ingresos de esas personas.

df10 = df.loc[df['Edad'].between(30, 40)].groupby('Nivel Educativo').agg(
    total_personas=('Nivel Educativo', 'size'),
    promedio_ingresos=('Ingresos', 'mean'))
display(df10)
```
