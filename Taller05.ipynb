```python
# Importar Pandas y guarda en la variable df el archivo Excel ubicado en: https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx
import pandas as pd
dfp = pd.read_excel("https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx")
df1 = display(pd.read_excel("https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx"))


```

```python
# Mostrar las primeras 10 filas de la columna que tiene el número de hijos

dfh = display(dfp["No de hijos"].head(10))
```

```python
# Mostar la cédula, nombre y edad de las primeras 5 personas del dataframe
dfc = display(dfp[['Cédula','Nombres','Edad']].head(5))
```

```python
# Presentar la cédula y estado civil de las últimas 10 personas
dfl = display(dfp[['Cédula','Estrato Socio']].tail(10))
```

```python
# Presentar las estadísticas de las columnas edad e ingresos de todas las personas del dataframe

dfs = display(dfp[['Cédula','Edad','Ingresos']].describe())
```

```python
# Presentar las estadísticas de las columnas peso y talla pero de solo
# las primeras 50 personas del dataframe
dfs2 = display(dfp[['Peso','Talla','Ingresos']].head(50).describe())
```

```python
# Muestra los datos de la persona ubicada en el índice o posición 100 del dataframe
# Puede usar la operación loc o iloc

dfs2 = display(dfp.iloc[100])
```

```python
# Muestre todas las columnas de las personas ubicadas en las posiciones de la 277 a la 286
dfi = display(dfp.loc[277:286])
```

```python
# Muestre solamente la cédula de la persona ubicada en la fila 183
dfi183 = display(dfp.loc[183, 'Cédula'])
```

```python
# Muestre la cedula, nombre y estado civil de las personas ubicadas entre las filas
# 417 y 423 del dataframe

df417 = display(dfp.loc[417:423, ['Cédula', 'Nombres', 'Estrato Socio']])


```

```python
# Muestre las estadísticas de la columna edad de las personas
# ubicadas entre las filas 300 y 399
dfi300 = display(dfp.loc[417:423, ['Edad']].describe())
```

```python
# Usando las operaciones de agregación (sección 11 del cheat sheer)
# Muestre la suma de los ingresos de todas las personas

dfa = display(dfp.Ingresos.sum())
```

```python
# Ahora muestre la suma de los ingresos de las personas
# ubicadas entre las filas 100 y 199 del data frames
dfa2 = display(dfp.loc[100:199, ['Ingresos']].sum())
```

```python
# Muestre el promedio de edad de las personas ubicadas
# entre las filas 200 y 250 del data frame
df3 = display(dfp.loc[200:250, ['Edad']].mean())
```

```python
# Determine la edad más grande (use la operación max)
# entre las personas ubicadas en las primeras 100 filas
df4 = display(dfp.loc[0:100, ['Edad']].max())
```

```python
# Encuentre el número de elementos únicos (use nunique)
# que se pueden encontrar dentro de la columna nivel educativo
df5 = display(dfp['Nivel Educativo'].nunique())
```

```python
# Encuentre la desviación estándar (use std) del peso y la talla
# de las personas ubicada entre la posición 300 y el final del dataframe
df4 = display(dfp.loc[300:, ['Peso','Talla']].std())
```
