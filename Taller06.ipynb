En este taller revisaremos las opciones que nos ofrece **pandas** para filtrar
datos basándonos en una determinada condición.

```python
# Vamos a seguir trabajando con el archivo de personas, así que en esta celda
# importamos pandas y guardamos en la variable df el archivo Excel ubicado en:
# https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx
import pandas as pd

df = pd.read_excel("https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx")
display(df)
```

```python
# Muestre las primeras 10 personas del dataframe
display(df.head(10))
```

```python
# Presente todas las columnas de aquellas personas cuyo peso sea superior a 90 kilos
personas_peso_mayor_90 = df.loc[df['Peso'] > 90]
display(personas_peso_mayor_90)
```

```python
# Muestre la cantidad de personas cuyo peso supere los 90 kilos
cantidad_peso_mayor_90 = df.loc[df['Peso'] > 90].shape[0]
display(cantidad_peso_mayor_90)
```

```python
# Muestre solamente la edad de las personas cuyo peso no supere los 80 kilos
edades_peso_hasta_80 = df.loc[df['Peso'] <= 80, 'Edad']
display(edades_peso_hasta_80)
```

```python
# Determine la cantidad de personas cuya edad esté por encima de los 50 años
cantidad_mayores_50 = df.loc[df['Edad'] > 50].shape[0]
display(cantidad_mayores_50)
```

```python
# Obtenga el promedio de la talla de las personas cuyos ingresos estén
# por encima del salario mínimo legal mensual vigente de Colombia para 2026
smlv = 1750905
promedio_talla_sobre_smlv = df.loc[df['Ingresos'] > smlv, 'Talla'].mean()
display(promedio_talla_sobre_smlv)
```

```python
# Encuentre el promedio de edad de las personas que no tienen hijos
promedio_edad_sin_hijos = df.loc[df['No de hijos'] == 0, 'Edad'].mean()
display(promedio_edad_sin_hijos)
```

```python
# Calcule el promedio de edad de las personas que fuman
promedio_edad_fumadores = df.loc[df['Fuma'] == 'SI', 'Edad'].mean()
display(promedio_edad_fumadores)
```

```python
# Calcule la suma de los ingresos de las personas que tienen automóvil
# entre las últimas 300 filas del dataframe
ultimas_300 = df.tail(300)
suma_ingresos_carro_ultimas_300 = ultimas_300.loc[
    ultimas_300['Tiene Automovil'] == 'SI',
    'Ingresos'
].sum()
display(suma_ingresos_carro_ultimas_300)
```

```python
# Cuente cuántas personas tienen un nombre que inicia por "A" mayúscula
cantidad_nombres_a = df.loc[df['Nombres'].str.startswith('A'), 'Nombres'].count()
display(cantidad_nombres_a)
```

```python
# Calcule el promedio del peso y la suma de la talla de las personas
# cuyo nivel educativo no es primaria
personas_no_primaria = df.loc[df['Nivel Educativo'] != 'PRIMARIA']
resultado_no_primaria = personas_no_primaria.agg({
    'Peso': 'mean',
    'Talla': 'sum'
})
display(resultado_no_primaria)
```

```python
# Encuentre la edad más pequeña entre las personas que no usan lentes
edad_minima_sin_lentes = df.loc[df['Usa Lentes'] == 'NO', 'Edad'].min()
display(edad_minima_sin_lentes)
```

```python
# Entre las últimas 100 personas, encuentre el ingreso más alto de quienes
# tienen dos hijos o más
ultimas_100 = df.tail(100)
ingreso_maximo_ultimas_100 = ultimas_100.loc[
    ultimas_100['No de hijos'] >= 2,
    'Ingresos'
].max()
display(ingreso_maximo_ultimas_100)
```

```python
# Calcule el promedio de la talla de las personas ubicadas entre las filas
# 100 y 200 que tengan una edad entre 30 y 40 años
filas_100_200 = df.loc[100:200]
promedio_talla_30_40 = filas_100_200.loc[
    filas_100_200['Edad'].between(30, 40),
    'Talla'
].mean()
display(promedio_talla_30_40)
```

```python
# Determine el promedio de ingresos de los hombres
promedio_ingresos_hombres = df.loc[df['Genero'] == 'HOMBRE', 'Ingresos'].mean()
display(promedio_ingresos_hombres)
```

```python
# Determine el promedio de ingresos de las mujeres
promedio_ingresos_mujeres = df.loc[df['Genero'] == 'MUJER', 'Ingresos'].mean()
display(promedio_ingresos_mujeres)
```

### **Respuesta**

El promedio de ingresos de los **hombres** es mayor:

- Hombres: **$3.768.316,79**
- Mujeres: **$3.540.369,65**

La diferencia promedio es de aproximadamente **$227.947,14** a favor de los hombres.


### **Reto final**

Están muy angustiados en la alcaldía porque parece que a mayor nivel educativo, hay más gordos. Presente la cantidad de gordos (personas que pesan más de 90 kilos) que hay en cada nivel
educativo y determine si es real o no la angustia del alcalde. Por cierto, peude usar la
operación `unique()` de una columna para saber los valores distintos de la columna del
nivel educativo.

```python
# Cuente las personas que pesan más de 90 kilos en cada nivel educativo
orden_niveles = ['PRIMARIA', 'SECUNDARIA', 'PREGRADO', 'POSTGRADO']
gordos_por_nivel = (
    df.loc[df['Peso'] > 90]
      .groupby('Nivel Educativo')
      .size()
      .reindex(orden_niveles, fill_value=0)
      .rename('Cantidad de personas con peso > 90 kg')
)
display(gordos_por_nivel)
```

#### **Conclusión del reto final**

La preocupación de la alcaldía **no está respaldada por la muestra**. Las cantidades de personas con peso superior a 90 kg son:

- Primaria: **24**
- Secundaria: **29**
- Pregrado: **25**
- Postgrado: **24**

La cantidad no aumenta de forma progresiva con el nivel educativo; secundaria presenta el valor más alto y postgrado vuelve a 24 personas.
