En este taller revisaremos las opciones que nos ofrece **pandas** para filtrar
datos basándonos en una determinada condición.

```python
# Vamos a seguir trabajando con el archivo de personas, así que en esta celda
# Importar Pandas y guarda en la variable df el archivo Excel
# ubicado en: https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx
import pandas as pd
dfp = pd.read_excel("https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx")
display(dfp)
```

```python
# Muestre las primeras 10 personas del data frame de personas
dfp.Nombres.head(10)
```

```python
# Comencemos con algo fácil: presente todas columnas de de aquellas personas
# cuyo peso sea superior a 90 kilos.
dfw = dfp[dfp.Peso > 90]
display(dfw)
```

```python
# Ahora quiero ver la cantidad de personas del dataframe cuyo peso
# supere los 90 kilos

df90 = dfp[dfp.Peso > 90][['Peso']].count()
display(df90)

#
```

```python
# ¿Será posible ver solamente la edad de las personas cuyo peso no supere los 80 kilos?
df80 = dfp[dfp.Peso < 80].Edad
display(df80)

```

```python
# Quiero ahora determinar la cantidad de personas cuya edad esté por encima de los 50 años
df50 = dfp[dfp.Edad > 50].Edad.count()
display(df50)
```

```python
# ¿Cómo sería para obtener el promedio de la talla de las personas cuyos ingresos estén
# por encima del salario mínimo legal vigente?

smlv = 2000000
dfta = dfp[dfp.Ingresos > smlv ].Talla.mean()
display(dfta)


```

```python
# Me encantaría ahora encontrar el promedio de edad de las personas que no tienen hijos
dfnh = dfp[dfp['No de hijos'] == 0]['Edad'].mean()
display(dfnh)
```

```python
# Necesito saber el promedio de edad de las personas que fuman
dff = dfp[dfp['Fuma'] == 'NO']['Edad'].mean()
display(dff)
```

```python
# Y veamos si son capaces de obtener la suma de los ingresos de las personas
# que tienen carro pero que están ubicadas entre las últimas 300 filas
dfc = dfp[dfp['Tiene Automovil'] == 'SI'][300:].Ingresos.sum()
display(dfc)
```

```python
# ¿Y cuántas personas cuyo nombre inicia por "A" en mayúsculas hay en todo el dataframes?
dfa = dfp[dfp.Nombres.str.startswith('A')].Nombres.count()
display(dfa)
```

```python
# Desearía saber el promedio del peso y la suma de la talla
# de las personas cuyo nivel educativo no es primaria
dfpa = dfp[dfp['Nivel Educativo'] != 'Primaria']
dfpaagg = dfpa.agg({
    'Peso': 'mean',
    'Talla': 'sum'
})
display(dfpaagg)
```

```python
# ¿Eres capaz de encontrar la edad más pequeña entre todas las
# personas  que no usan lentes?

dfem = dfp[dfp['Usa Lentes'] == 'SI']['Edad'].min()

display(dfem)

```

```python
# Y entre las últimas 100 personas, ¿Cuál es el Ingreso más alto de
# las personas que tienen 2 hijos o más?

dfia = dfp[dfp['No de hijos'] >=2 ].tail(100)['Ingresos'].max()

display(dfia)
```

```python
# Ahora tengo el capricho de saber si hay forma de determina el promedio de la talla
# de todas las personas ubicadas entre las filas 100 y 200, pero que tengan
# una edad entre 30 y 40 años (puede usar el between)
dfia1 = dfp.iloc[100:200]
dfia2 = dfia1[dfp['Edad'].between(30, 40) ]['Talla'].mean()
display(dfia2)
```

```python
# ¿Será que puede determinarme ahora el promedio de ingresos de los hombres?
dfih = dfp[dfp['Genero'] == 'HOMBRE' ]['Ingresos'].mean()
display(dfih)
```

```python
# Y ahora quiero saber el promedio de los ingresos de las mujeres?
dfim = dfp[dfp['Genero'] == 'MUJER' ]['Ingresos'].mean()
display(dfim)
```
