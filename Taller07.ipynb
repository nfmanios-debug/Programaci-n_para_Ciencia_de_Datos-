En este taller revisaremos las opciones que nos ofrece **pandas** para filtrar
datos basándonos en una determinada condición.

```python
# Vamos a seguir trabajando con el archivo de personas, así que en esta celda
# importamos pandas y guardamos el archivo Excel en la variable df1:
# https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx
import pandas as pd

df1 = pd.read_excel("https://github.com/luiscobo/poo/raw/refs/heads/main/Personas.xlsx")
display(df1)
```

```python
# Determine cuántas mujeres mayores de 30 años no fuman
df2 = df1.loc[
    (df1['Genero'] == 'MUJER') &
    (df1['Edad'] > 30) &
    (df1['Fuma'] == 'NO'),
    'Genero'
].count()
display(df2)
```

```python
# Muestre la cantidad de personas que viven en el estrato 2 o en el estrato 3
dfe2 = df1.loc[
    (df1['Estrato Socio'] == 2) |
    (df1['Estrato Socio'] == 3),
    'Estrato Socio'
].count()
display(dfe2)
```

```python
# Muestre los nombres de los hombres con talla entre 167 y 173 cm
# y edad entre 50 y 57 años
dfe3 = df1.loc[
    (df1['Genero'] == 'HOMBRE') &
    (df1['Talla'].between(167, 173)) &
    (df1['Edad'].between(50, 57)),
    'Nombres'
]
display(dfe3)
```

```python
# Calcule el promedio del peso de las mujeres sin hijos, sin carro y sin casa
dfavg = df1.loc[
    (df1['Genero'] == 'MUJER') &
    (df1['No de hijos'] == 0) &
    (df1['Tiene Casa'] == 'NO') &
    (df1['Tiene Automovil'] == 'NO'),
    'Peso'
].mean()
display(dfavg)
```

```python
# Calcule la suma de los ingresos de los hombres que tienen nivel educativo
# de primaria y que fuman
dfhep = df1.loc[
    (df1['Genero'] == 'HOMBRE') &
    (df1['Nivel Educativo'] == 'PRIMARIA') &
    (df1['Fuma'] == 'SI'),
    'Ingresos'
].sum()
display(dfhep)
```

```python
# Cuente las personas que no tienen casa ni carro y cuyos ingresos
# son superiores a tres millones de pesos
dfnc3 = df1.loc[
    (df1['Tiene Casa'] == 'NO') &
    (df1['Tiene Automovil'] == 'NO') &
    (df1['Ingresos'] > 3000000),
    'Ingresos'
].count()
display(dfnc3)
```

```python
# Imprima la edad más grande y luego los nombres de las personas que tienen esa edad
dfmax = df1['Edad'].max()
display(dfmax)

display(df1.loc[df1['Edad'] == dfmax, 'Nombres'])
```
