# Expresiones Lógicas

Las expresiones aritméticas, al evaluarse, producen números como resultados; una expresión lógica o _Booleana_ (algunas veces llamadas _predicados_) generan como resultados algunos de los siguientes valores de verdad: ```True``` (**Verdadero** o **Cierto**) y ```False``` ( **Falso**). Estas expresiones las usaremos en los programas ya sea para las instrucciones condicionales o repetitivas.

## Operadores de Comparación
Estos son operadores que dan como resultado un valor de verdad o valor lógico. Estos operadores se pueden pensar como una pregunta y al responder tal pregunta se obtiene un ```True``` o un ```False``` como respuesta. Muchos de estos operadores son conocidos por todos nosotros y en Python son los siguientes:

Operador | Descripción | Ejemplo
---|---|---
```==``` | Permita preguntar si dos elementos son iguales | ```3 == 3```
```!=```| Permite determinar si dos elementos son diferentes | ```2 != 3```
```<``` | Para preguntar si el primer elemento es inferior al segundo | ```2 < 3```
```>``` | Para saber si el primer elemento es superior al segundo | ```3 > 2```
```<= ```| Para preguntar si el primero elemento es menor _o igual_ al segundo | ```-3 <= 3```
```>= ```| Para preguntar si el primero elemento es mayor _o igual_ al segundo | ```8 >= 3```

### Ejercicio 1
Si tenemos que $a = 3, b = 5$ y $c = 7$, escriba frente a cada uno de las pruebas a continuación cuál es el valor de verdad que obtendríamos:

* $ (a+b) < c?$
* $ (a - b) < c?$
* $( a - b) = c?$
* $(a * b )\neq c$

````python
# Solución al Ejercicio 1

a = 3 ; b = 5 ; c = 7

assert ((a + b) < c) ==  False

assert ((a - b) < c) == True

assert ((a-b) == c) == False

assert ((a*b) != c) == True
````

## Operadores Lógicos

Adicionalmente a los operadores relacionales, Python cuenta con los operadores lógicos. Estos operadores se usan en combinación con los operadores relacionales para crear condiciones màs complejas. No deben ser desconocidas para nosotros, ya que las manejamos en nuestra vida diaria sin ningún tipo de inconveniente. Por ejemplo, cuando estamos indecisos en tomar para postre un helado _o_ un flan.

Hay tres operadores lógicos en Python, los cuáles podemos ver a continuación:

Operador | Descripción | Ejemplo
---|---|---
```and``` | Resulta en ```True``` si tanto la parte izquierda como la derecha son ciertos | ```(3 < 4) and (5 > 3)```
```or``` | Resulta en ```False``` si tanto la parte izquierda como la derecha son falsos | ```(3 < 4) or (3 > 5)```
```not``` | Resulta en ```True``` si el valor a evaluar es falso, y viceversa | ```not 3 < 2```

### Ejercicio 29

### Ejercicio 31

Debe preguntarle a un estudiantes las 3 notas que obtuvo en una materia, luego debe calcular y mostrar el promedio de las tres notas. Adicionalmente, imprimir si el alumno la palabra "REBROBO" si el promedio es inferior a 60.0, o imprima la palabra "EN VILO" si el promedio está entre 60.0 y 70.0, y finalmente imprima la palabra "APROBÓ", cuando la nota está por encima o es igual a 70.0.

````python
# Escriba aquí el programa

nota1 = float(input("Entre la primera nota: "))
nota2 = float(input("Entre la segunda nota: "))
nota3 = float(input("Entre la tercera nota: "))

promedio = (nota1 + nota2 + nota3) / 3
print(f"El promedio es: {promedio:.1f}")

if promedio < 60.0:
    print("REPROBÓ")
elif promedio >= 60.0 and promedio < 70.0:
    print("EN VILO")
else:
    print("APROBÓ")
````
